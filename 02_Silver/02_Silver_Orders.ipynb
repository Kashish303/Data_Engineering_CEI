```python
orders_df = orders_df.withColumn(
    "order_status",
    F.trim(F.lower(F.col("order_status")))
)
```

```python
# Delivery duration 
orders_df = orders_df.withColumn(
    "delivery_duration_days",
    F.when(
        F.col("order_delivered_customer_date").isNotNull(),
        F.datediff(
            F.col("order_delivered_customer_date"),
            F.col("order_purchase_timestamp")
        )
    )
)
```

```python
# Business Rule Validation

# 1: If an order is delivered,then the delivery date cannot be NULL.
# 2: If the order is NOT delivered,then why is there already a delivery date?
# 3: Customer receive the package before buying it
# 4: The system promised delivery before the customer even ordered.

#invalid_business_rules are records that failed business logic.

invalid_business_rules = (
    orders_df.filter(
        ((F.col("order_status") == "delivered") &F.col("order_delivered_customer_date").isNull())
        |
        ((F.col("order_status") != "delivered") &F.col("order_delivered_customer_date").isNotNull())
        |
        (F.col("order_delivered_customer_date").isNotNull() &(F.col("order_delivered_customer_date") <F.col("order_purchase_timestamp")))
        |
        (F.col("order_estimated_delivery_date") <F.col("order_purchase_timestamp"))
    )
    .withColumn(
        "quarantine_reason",
        F.when((F.col("order_status") == "delivered") &F.col("order_delivered_customer_date").isNull(),"Delivered Order Missing Delivery Date")
        .when((F.col("order_status") != "delivered") &F.col("order_delivered_customer_date").isNotNull(),"Invalid Delivery Date for Non-Delivered Order")
        .when(F.col("order_delivered_customer_date") <F.col("order_purchase_timestamp"), "Delivery Before Purchase")
        .when(F.col("order_estimated_delivery_date") <F.col("order_purchase_timestamp"),"Estimated Delivery Before Purchase")
    )
)
print(f"Business Rule Violations : {invalid_business_rules.count()}")
```

```python
# Foreign Key Validation
# orphan_orders are records whose customer_id does not exist.

silver_customers = spark.table(SILVER_CUSTOMERS)

orphan_orders = (orders_df.join(
silver_customers.select("customer_id"),
on="customer_id",
how="left_anti"
).withColumn(
"quarantine_reason",
F.lit("Customer ID Not Found")
    )
)

print(f"Orphan Orders : {orphan_orders.count()}")
```

```python
# Invalid Orders
invalid_orders = (
    invalid_business_rules
    .unionByName(orphan_orders)
    .dropDuplicates(["order_id"])
)

print(f"Total Invalid Orders : {invalid_orders.count()}")
```

```python
# Valid Orders

valid_orders = (orders_df.join(
invalid_orders.select("order_id"),
on="order_id",
how="left_anti"
    )
)
print(f"Valid Orders : {valid_orders.count()}")
```

```python
# Refresh Audit Columns
valid_orders = add_audit_columns(valid_orders,PIPELINE_NAME,RUN_ID)
```

```python
# Write qurantine table 
(
    invalid_orders.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(QUARANTINE_ORDERS)
)
```

```python
from delta.tables import DeltaTable
table_exists = spark.catalog.tableExists(TARGET_TABLE)

if not table_exists:

    (
        valid_orders.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(TARGET_TABLE)
    )

    status = "CREATED"

else:

    delta_table = DeltaTable.forName(
        spark,
        TARGET_TABLE
    )

    (
        delta_table.alias("target")
        .merge(
            valid_orders.alias("source"),
            "target.order_id = source.order_id"
        )
        .whenMatchedUpdateAll()
        .whenNotMatchedInsertAll()
        .execute()
    )

    status = "MERGED"

pipeline_status = "SUCCESS"
```

```python
silver_orders = spark.table(TARGET_TABLE)
rows_written = silver_orders.count()
display(silver_orders.limit(10))

distinct_order_ids = validate_primary_key(
    orders_df,
    "order_id"
)
```
